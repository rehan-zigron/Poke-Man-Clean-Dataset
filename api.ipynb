```python
# Abilities ko readable banao
df['abilities_clean'] = df['abilities'].apply(
    lambda x: ", ".join([a['ability']['name'] for a in x])
)

# Types ko readable banao
df['types_clean'] = df['types'].apply(
    lambda x: ", ".join([t['type']['name'] for t in x])
)

# Stats ko readable banao
df['stats_clean'] = df['stats'].apply(
    lambda x: ", ".join([f"{s['stat']['name']}:{s['base_stat']}" for s in x])
)

df[['name', 'height', 'weight', 'base_experience', 'abilities_clean', 'types_clean', 'stats_clean']]
```

```python
clean_df = df[['name', 'height', 'weight', 'base_experience', 'abilities_clean', 'types_clean', 'stats_clean']]
clean_df.to_csv("pokemon_clean.csv", index=False)
print("Saved!")
```

## LOOP DATA FETCHING 

```python
import requests
import pandas as pd

all_pokemon = []   # khaali list — har pokemon ka data isme jayega

for i in range(1, 51):   # 1 se 10 pokemon (baad mein 51 kar sakte ho)
    url = f"https://pokeapi.co/api/v2/pokemon/{i}"
    response = requests.get(url)
    r = response.json()
    
    all_pokemon.append({
        'id': r['id'],
        'name': r['name'],
        'height': r['height'],
        'weight': r['weight'],
        'base_experience': r['base_experience'],
        'types': ", ".join([t['type']['name'] for t in r['types']]),
        'abilities': ", ".join([a['ability']['name'] for a in r['abilities']]),
    })

df_all = pd.DataFrame(all_pokemon)
df_all
```
